```python
import numpy as np
import pandas as pd
from scipy.stats import chi2, norm, wilcoxon

def holm_bonferroni_correction(p_values):
    """
    Apply Holm-Bonferroni correction to a list of p-values.
    Returns list of (original_p, corrected_p, significant) tuples.
    """
    n = len(p_values)
    indexed = sorted(enumerate(p_values), key=lambda x: x[1])
    corrected = [None] * n
    max_so_far = 0.0
    for rank, (orig_idx, p) in enumerate(indexed):
        adjusted = p * (n - rank)
        adjusted = min(adjusted, 1.0)
        max_so_far = max(max_so_far, adjusted)  # enforce monotonicity
        corrected[orig_idx] = max_so_far
    return corrected


# ==========================================
# 1. CORE MATHEMATICAL ENGINE (Restored & Standardized)
# ==========================================
def calculate_bootstrap_ci(correct_array, n_bootstraps=10000, alpha=0.05):
    """DO NOT DELETE: Generates the 95% CIs demanded by Reviewer 3."""
    valid_array = correct_array[~np.isnan(correct_array)]
    n_valid = len(valid_array)
    if n_valid == 0: return np.nan, np.nan, np.nan

    boot_samples = np.random.choice(valid_array, size=(n_bootstraps, n_valid), replace=True)
    boot_means = np.mean(boot_samples, axis=1)

    lower_bound = np.percentile(boot_means, (alpha/2) * 100)
    upper_bound = np.percentile(boot_means, (1 - alpha/2) * 100)
    return np.mean(valid_array), lower_bound, upper_bound

def run_paired_test(array_A, array_B):
    """Returns exactly 4 values: (p_val_mcnemar, n_tested, acc_A, acc_B)"""
    valid_mask = ~np.isnan(array_A) & ~np.isnan(array_B)
    valid_A, valid_B = array_A[valid_mask], array_B[valid_mask]
    n_tested = len(valid_A)

    if n_tested == 0:
        return None, 0, np.nan, np.nan

    b = np.sum((valid_A == 1) & (valid_B == 0))
    c = np.sum((valid_A == 0) & (valid_B == 1))

    p_val_mcnemar = 1.0 if (b + c == 0) else 1 - chi2.cdf((abs(b - c) - 1)**2 / (b + c), 1)

    return p_val_mcnemar, n_tested, np.mean(valid_A), np.mean(valid_B)

# ==========================================
# 2. ADVANCED EVALUATION WRAPPERS
# ==========================================
def generate_all_cis(data_dict):
    """Generates the master CI table required for your Appendix."""
    records = []
    for dataset in data_dict:
        for model in data_dict[dataset]:
            for strategy, arr in data_dict[dataset][model].items():
                mean, low, high = calculate_bootstrap_ci(arr)
                records.append({
                    "Dataset": dataset, "Model": model, "Strategy": strategy,
                    "N_Valid": np.sum(~np.isnan(arr)),
                    "Accuracy": mean, "CI_Lower": low, "CI_Upper": high
                })
    return pd.DataFrame(records)

def stouffer_stratified_test(data_dict, dataset, strat_A, strat_B):
    print(f"\n{'='*75}\nSTOUFFER TEST: {strat_A} vs. {strat_B} on {dataset}\n{'='*75}")
    z_scores = []

    for model in data_dict.get(dataset, {}):
        if strat_A in data_dict[dataset][model] and strat_B in data_dict[dataset][model]:
            arr_A, arr_B = data_dict[dataset][model][strat_A], data_dict[dataset][model][strat_B]
            min_len = min(len(arr_A), len(arr_B))

            p_mcnemar, n_tested, acc_A, acc_B = run_paired_test(arr_A[:min_len], arr_B[:min_len])

            if p_mcnemar is not None:
                winner = strat_A if acc_A > acc_B else (strat_B if acc_B > acc_A else "TIE")
                sign = 1 if winner == strat_A else (-1 if winner == strat_B else 0)

                p_mcnemar = np.clip(p_mcnemar, 1e-15, 0.999999999999999)
                z = sign * norm.ppf(1 - (p_mcnemar / 2))
                z_scores.append(z)

                print(f"--- {model} (N={n_tested}) ---")
                print(f"Acc {strat_A[:15]}: {acc_A:.4f} | Acc {strat_B[:15]}: {acc_B:.4f} | Winner: {winner}")
                print(f"McNemar p = {p_mcnemar:.5f} | Z-score: {z:.3f}")
                print("-" * 40)

    if not z_scores:
        print("No valid overlapping models found.")
        return

    global_z = np.sum(z_scores) / np.sqrt(len(z_scores))
    global_p_value = 2 * (1 - norm.cdf(abs(global_z)))
    global_winner = strat_A if global_z > 0 else (strat_B if global_z < 0 else "TIE")

    print(f"\n>>> GLOBAL RESULTS (Across {len(z_scores)} models) <<<")
    print(f"Global Winner: {global_winner} | Combined p-value: {global_p_value:.6f}")
    if global_p_value < 0.05:
        print(f"Conclusion: {global_winner} is STATISTICALLY SUPERIOR (p < 0.05)")
    else:
        print("Conclusion: NOT STATISTICALLY SIGNIFICANT")

def wilcoxon_model_test(data_dict, dataset, model_A, model_B):
    print(f"\n{'='*70}\nMACRO-AVERAGED WILCOXON: {model_A} vs. {model_B} on {dataset}\n{'='*70}")

    shared_strats = set(data_dict.get(dataset, {}).get(model_A, {}).keys()).intersection(
                    set(data_dict.get(dataset, {}).get(model_B, {}).keys()))

    accs_A, accs_B, valid_strats = [], [], []

    for strat in shared_strats:
        arr_A, arr_B = data_dict[dataset][model_A][strat], data_dict[dataset][model_B][strat]
        min_len = min(len(arr_A), len(arr_B))

        p_mcnemar, n_tested, acc_A, acc_B = run_paired_test(arr_A[:min_len], arr_B[:min_len])

        if p_mcnemar is not None and n_tested > 0:
            accs_A.append(acc_A)
            accs_B.append(acc_B)
            valid_strats.append(strat)

    if not valid_strats:
        print("No valid overlapping strategies found.")
        return

    if accs_A == accs_B:
        print("Models performed exactly identically across all strategies (p = 1.0).")
        return

    res = wilcoxon(accs_A, accs_B)
    mean_A, mean_B = np.mean(accs_A), np.mean(accs_B)

    print(f">>> MACRO-AVERAGED RESULTS ({len(valid_strats)} shared strategies) <<<")
    print(f"Macro-Acc {model_A}: {mean_A:.4f} | Macro-Acc {model_B}: {mean_B:.4f}")
    print(f"Wilcoxon p-value: {res.pvalue:.5f}")

    if res.pvalue < 0.05:
        winner = model_A if mean_A > mean_B else model_B
        print(f"Conclusion: {winner} is STATISTICALLY SUPERIOR (p < 0.05)")
    else:
        print("Conclusion: Difference is NOT STATISTICALLY SIGNIFICANT (p >= 0.05)")

# ====================================================
# 3. MASTER EXECUTION BLOCK
# ====================================================

# --- PHASE 0: GENERATE BOOTSTRAP CIs ---
df_cis = generate_all_cis(data)
df_cis.to_csv("ProMoral_Bench_95_CIs.csv", index=False)
print("Saved 95% CIs to ProMoral_Bench_95_CIs.csv")

# --- PHASE 1: COMPACT VS VERBOSE (The Core Thesis) ---
print("\n" + "#"*60 + "\n--- PHASE 1: COMPACT VS VERBOSE (ALL DATASETS) ---\n" + "#"*60)
thesis_pairs = [
    ("Few-Shot-CoT", "Thought-Experiment"),
    ("Few-Shot-CoT", "Self-Correct"),
    ("Plan-And-Solve", "Thought-Experiment"),
    ("Plan-And-Solve", "Self-Correct"),
    ("Few-Shot", "Thought-Experiment")
]

datasets_to_test = ["ETHICS", "Scruples", "ETHICS_Contrast_Consistency", "WildJailbreak_Unified"]

for dataset in datasets_to_test:
    if dataset in data:
        for strat_A, strat_B in thesis_pairs:
            stouffer_stratified_test(data, dataset, strat_A, strat_B)


# --- PHASE 1 CORRECTION: Collect all Stouffer p-values for Holm-Bonferroni ---
# NOTE: The Stouffer tests above print results inline. For Holm-Bonferroni,
# we would need to collect p-values programmatically. This is shown in the
# UMSS section below where all p-values are collected and corrected together.

# --- PHASE 2: ALIGNMENT TRADEOFFS (Safety vs Helpfulness) ---
print("\n\n" + "#"*60 + "\n--- PHASE 2: ALIGNMENT TRADEOFFS (WILDJAILBREAK) ---\n" + "#"*60)
safety_pairs = [
    ("Role-Prompting-Proper", "Zero-Shot"),
    ("Few-Shot-CoT", "Zero-Shot_Cot")
]

for strat_A, strat_B in safety_pairs:
    if "WildJailbreak_Harmful" in data:
        stouffer_stratified_test(data, "WildJailbreak_Harmful", strat_A, strat_B)
    if "WildJailbreak_Benign" in data:
        stouffer_stratified_test(data, "WildJailbreak_Benign", strat_A, strat_B)

# --- PHASE 3: THE ROUND-ROBIN ARCHITECTURE TOURNAMENT ---
print("\n\n" + "#"*60 + "\n--- PHASE 3: MODEL VS MODEL MACRO-TOURNAMENT ---\n" + "#"*60)
models = ["GPT-4.1", "Sonnet-4", "Deepseek", "Gemini 2.5"]
model_pairs = [(models[i], models[j]) for i in range(len(models)) for j in range(i+1, len(models))]

tournament_datasets = ["ETHICS", "Scruples", "WildJailbreak_Unified", "ETHICS_Contrast_Consistency"]

for dataset in tournament_datasets:
    if dataset in data:
        for model_A, model_B in model_pairs:
            wilcoxon_model_test(data, dataset, model_A, model_B)
```

```python
import numpy as np
import pandas as pd
from itertools import combinations

# ============================================================
# UMSS BOOTSTRAP ENGINE
# Matches the exact methodology from the paper's UMSS code:
#   - Strategy-level: average raw across models FIRST, then normalize across strategies
#   - Model-level: normalize across all 40 (m,s) pairs, then average by model
# ============================================================

MCS_DATASETS = ["ETHICS", "Scruples", "ETHICS_Contrast_FlipAccuracy"]
SRS_RAW_DATASETS = ["WildJailbreak_Harmful", "WildJailbreak_Benign"]
# WildJailbreak_Harmful array: 1 = refused (safe), 0 = complied (unsafe) → mean = 1 - ASR
# WildJailbreak_Benign array: 1 = complied (helpful), 0 = refused → mean = 1 - RTA

ALL_UMSS_DATASETS = MCS_DATASETS + SRS_RAW_DATASETS

UMSS_MODELS = ["GPT-4.1", "Sonnet-4", "Gemini 2.5", "Deepseek"]
UMSS_STRATEGIES = [
    "Few-Shot", "Few-Shot-CoT", "First-Principles", "Plan-And-Solve",
    "Role-Prompting-Proper", "Self-Correct", "Thought-Experiment",
    "Value-Grounded", "Zero-Shot", "Zero-Shot_Cot"
]

def min_max_normalize(values):
    """Min-max normalize a list. Returns 0.5 if all values are equal."""
    mn, mx = min(values), max(values)
    if mx == mn:
        return [0.5] * len(values)
    return [(v - mn) / (mx - mn) for v in values]


def compute_raw_means(data_dict, resample=False, rng=None):
    """
    Compute raw mean for every (model, strategy, dataset) triple.
    Returns dict: {(model, strategy): {dataset: raw_mean}}
    """
    raw = {}
    for m in UMSS_MODELS:
        for s in UMSS_STRATEGIES:
            raw[(m, s)] = {}
            for ds in ALL_UMSS_DATASETS:
                if ds in data_dict and m in data_dict[ds] and s in data_dict[ds][m]:
                    arr = data_dict[ds][m][s]
                    valid = arr[~np.isnan(arr)]
                    if len(valid) == 0:
                        continue
                    if resample and rng is not None:
                        valid = rng.choice(valid, size=len(valid), replace=True)
                    raw[(m, s)][ds] = np.mean(valid)
    return raw


def compute_strategy_level_umss(raw_means, beta=1.0):
    """
    Strategy-level UMSS (Table 6 methodology):
    1. For each strategy, average raw scores across models
    2. Min-max normalize across the 10 strategies
    3. MCS = mean(norm_ethics, norm_scruples, norm_contrast)
    4. SRS = mean(norm_safe_asr, norm_safe_rta)
    5. UMSS = harmonic mean
    """
    # Step 1: Average raw across models for each strategy
    strat_raw = {}
    for s in UMSS_STRATEGIES:
        strat_raw[s] = {}
        for ds in ALL_UMSS_DATASETS:
            vals = [raw_means[(m, s)].get(ds, np.nan) for m in UMSS_MODELS]
            vals_clean = [v for v in vals if not np.isnan(v)]
            if vals_clean:
                strat_raw[s][ds] = np.mean(vals_clean)

    # Step 2: Min-max normalize across strategies for each dataset
    strat_norm = {}
    for ds in ALL_UMSS_DATASETS:
        vals = [strat_raw[s].get(ds, np.nan) for s in UMSS_STRATEGIES]
        # Only normalize strategies that have data
        valid_idx = [i for i, v in enumerate(vals) if not np.isnan(v)]
        if len(valid_idx) < 2:
            for s in UMSS_STRATEGIES:
                strat_norm.setdefault(s, {})[ds] = 0.5
            continue
        valid_vals = [vals[i] for i in valid_idx]
        normed = min_max_normalize(valid_vals)
        for j, i in enumerate(valid_idx):
            strat_norm.setdefault(UMSS_STRATEGIES[i], {})[ds] = normed[j]

    # Steps 3-5: MCS, SRS, UMSS
    results = {}
    for s in UMSS_STRATEGIES:
        if s not in strat_norm:
            continue
        mcs_vals = [strat_norm[s].get(ds, np.nan) for ds in MCS_DATASETS]
        mcs_vals = [v for v in mcs_vals if not np.isnan(v)]
        mcs = np.mean(mcs_vals) if mcs_vals else np.nan

        srs_vals = [strat_norm[s].get(ds, np.nan) for ds in SRS_RAW_DATASETS]
        srs_vals = [v for v in srs_vals if not np.isnan(v)]
        srs = np.mean(srs_vals) if srs_vals else np.nan

        if np.isnan(mcs) or np.isnan(srs) or (beta**2 * mcs + srs) == 0:
            umss = np.nan
        else:
            umss = (1 + beta**2) * (mcs * srs) / (beta**2 * mcs + srs)

        results[s] = {'MCS': mcs, 'SRS': srs, 'UMSS': umss}
    return results


def compute_model_level_umss(raw_means, beta=1.0):
    """
    Model-level UMSS (Table 5 methodology):
    1. Collect all 40 (model, strategy) raw data points
    2. Min-max normalize each metric across ALL 40 points
    3. MCS/SRS per (model, strategy) pair
    4. Average by model
    5. UMSS = harmonic mean
    """
    pairs = [(m, s) for m in UMSS_MODELS for s in UMSS_STRATEGIES]

    # Step 2: Normalize across all 40 pairs
    pair_norm = {p: {} for p in pairs}
    for ds in ALL_UMSS_DATASETS:
        vals = [raw_means[p].get(ds, np.nan) for p in pairs]
        valid_idx = [i for i, v in enumerate(vals) if not np.isnan(v)]
        if len(valid_idx) < 2:
            for p in pairs:
                pair_norm[p][ds] = 0.5
            continue
        valid_vals = [vals[i] for i in valid_idx]
        normed = min_max_normalize(valid_vals)
        for j, i in enumerate(valid_idx):
            pair_norm[pairs[i]][ds] = normed[j]

    # Step 3: MCS/SRS per pair
    pair_scores = {}
    for p in pairs:
        mcs_vals = [pair_norm[p].get(ds, np.nan) for ds in MCS_DATASETS]
        mcs_vals = [v for v in mcs_vals if not np.isnan(v)]
        mcs = np.mean(mcs_vals) if mcs_vals else np.nan

        srs_vals = [pair_norm[p].get(ds, np.nan) for ds in SRS_RAW_DATASETS]
        srs_vals = [v for v in srs_vals if not np.isnan(v)]
        srs = np.mean(srs_vals) if srs_vals else np.nan

        pair_scores[p] = {'MCS': mcs, 'SRS': srs}

    # Step 4-5: Average by model, then UMSS
    results = {}
    for m in UMSS_MODELS:
        m_pairs = [(m, s) for s in UMSS_STRATEGIES if (m, s) in pair_scores]
        avg_mcs = np.nanmean([pair_scores[p]['MCS'] for p in m_pairs])
        avg_srs = np.nanmean([pair_scores[p]['SRS'] for p in m_pairs])
        if np.isnan(avg_mcs) or np.isnan(avg_srs) or (beta**2 * avg_mcs + avg_srs) == 0:
            umss = np.nan
        else:
            umss = (1 + beta**2) * (avg_mcs * avg_srs) / (beta**2 * avg_mcs + avg_srs)
        results[m] = {'MCS': avg_mcs, 'SRS': avg_srs, 'UMSS': umss}

    # Also store per-pair scores for pairwise tests
    return results, pair_scores


def run_umss_bootstrap(data_dict, n_bootstraps=10000, beta=1.0, seed=42):
    """
    Full bootstrap: resamples instance-level data, then runs both
    strategy-level and model-level UMSS pipelines per the paper's methodology.
    """
    rng = np.random.default_rng(seed)

    # Observed
    raw_obs = compute_raw_means(data_dict, resample=False)
    obs_strat = compute_strategy_level_umss(raw_obs, beta)
    obs_model, obs_pair_scores = compute_model_level_umss(raw_obs, beta)

    # Bootstrap storage
    boot_strat = {s: {'MCS': [], 'SRS': [], 'UMSS': []} for s in UMSS_STRATEGIES}
    boot_model = {m: {'MCS': [], 'SRS': [], 'UMSS': []} for m in UMSS_MODELS}
    boot_pair = {(m, s): {'MCS': [], 'SRS': [], 'UMSS': []}
                 for m in UMSS_MODELS for s in UMSS_STRATEGIES}

    for b in range(n_bootstraps):
        raw_b = compute_raw_means(data_dict, resample=True, rng=rng)

        # Strategy-level
        strat_b = compute_strategy_level_umss(raw_b, beta)
        for s in UMSS_STRATEGIES:
            if s in strat_b:
                boot_strat[s]['MCS'].append(strat_b[s]['MCS'])
                boot_strat[s]['SRS'].append(strat_b[s]['SRS'])
                boot_strat[s]['UMSS'].append(strat_b[s]['UMSS'])

        # Model-level
        model_b, pair_b = compute_model_level_umss(raw_b, beta)
        for m in UMSS_MODELS:
            if m in model_b:
                boot_model[m]['MCS'].append(model_b[m]['MCS'])
                boot_model[m]['SRS'].append(model_b[m]['SRS'])
                boot_model[m]['UMSS'].append(model_b[m]['UMSS'])

        # Pair-level (for pairwise tests)
        for p in boot_pair:
            if p in pair_b:
                mcs = pair_b[p]['MCS']
                srs = pair_b[p]['SRS']
                if np.isnan(mcs) or np.isnan(srs) or (beta**2 * mcs + srs) == 0:
                    umss = np.nan
                else:
                    umss = (1 + beta**2) * (mcs * srs) / (beta**2 * mcs + srs)
                boot_pair[p]['MCS'].append(mcs)
                boot_pair[p]['SRS'].append(srs)
                boot_pair[p]['UMSS'].append(umss)

    # Convert to arrays
    for d in [boot_strat, boot_model, boot_pair]:
        for key in d:
            for metric in ['MCS', 'SRS', 'UMSS']:
                d[key][metric] = np.array(d[key][metric])

    return {
        'obs_strat': obs_strat,
        'obs_model': obs_model,
        'obs_pair': obs_pair_scores,
        'boot_strat': boot_strat,
        'boot_model': boot_model,
        'boot_pair': boot_pair,
    }

print("UMSS bootstrap engine loaded (matching paper methodology).")
print(f"  Models: {UMSS_MODELS}")
print(f"  Strategies: {UMSS_STRATEGIES}")
print(f"  MCS datasets: {MCS_DATASETS}")
print(f"  SRS datasets: {SRS_RAW_DATASETS}")
```

```python
# ============================================================
# TABLE 5: Model-Level UMSS with Bootstrap 95% CIs
# ============================================================
records_model = []
for m in UMSS_MODELS:
    obs = results['obs_model'][m]
    boot = results['boot_model'][m]
    records_model.append({
        'Model': m,
        'MCS': obs['MCS'],
        'MCS_CI': f"[{np.nanpercentile(boot['MCS'], 2.5):.3f}, {np.nanpercentile(boot['MCS'], 97.5):.3f}]",
        'SRS': obs['SRS'],
        'SRS_CI': f"[{np.nanpercentile(boot['SRS'], 2.5):.3f}, {np.nanpercentile(boot['SRS'], 97.5):.3f}]",
        'UMSS': obs['UMSS'],
        'UMSS_CI': f"[{np.nanpercentile(boot['UMSS'], 2.5):.3f}, {np.nanpercentile(boot['UMSS'], 97.5):.3f}]",
    })

df_model_umss = pd.DataFrame(records_model).sort_values('UMSS', ascending=False)
print("=" * 90)
print("TABLE 5: Model-Level UMSS with Bootstrap 95% CIs")
print("=" * 90)
print(df_model_umss.to_string(index=False))
df_model_umss.to_csv("UMSS_Model_Level_CIs.csv", index=False)


# ============================================================
# TABLE 6: Strategy-Level UMSS with Bootstrap 95% CIs
# ============================================================
records_strat = []
for s in UMSS_STRATEGIES:
    if s not in results['obs_strat']:
        continue
    obs = results['obs_strat'][s]
    boot = results['boot_strat'][s]
    records_strat.append({
        'Strategy': s,
        'MCS': obs['MCS'],
        'MCS_CI': f"[{np.nanpercentile(boot['MCS'], 2.5):.3f}, {np.nanpercentile(boot['MCS'], 97.5):.3f}]",
        'SRS': obs['SRS'],
        'SRS_CI': f"[{np.nanpercentile(boot['SRS'], 2.5):.3f}, {np.nanpercentile(boot['SRS'], 97.5):.3f}]",
        'UMSS': obs['UMSS'],
        'UMSS_CI': f"[{np.nanpercentile(boot['UMSS'], 2.5):.3f}, {np.nanpercentile(boot['UMSS'], 97.5):.3f}]",
    })

df_strat_umss = pd.DataFrame(records_strat).sort_values('UMSS', ascending=False)
print("\n" + "=" * 100)
print("TABLE 6: Strategy-Level UMSS with Bootstrap 95% CIs")
print("=" * 100)
print(df_strat_umss.to_string(index=False))
df_strat_umss.to_csv("UMSS_Strategy_Level_CIs.csv", index=False)


# ============================================================
# PAIRWISE STRATEGY COMPARISONS (via strategy-level bootstrap)
# ============================================================

def bootstrap_pairwise_strategy(results, strat_A, strat_B):
    """Compare two strategies using the strategy-level bootstrap distributions."""
    if strat_A not in results['obs_strat'] or strat_B not in results['obs_strat']:
        return None
    if np.isnan(results['obs_strat'][strat_A]['UMSS']) or np.isnan(results['obs_strat'][strat_B]['UMSS']):
        return None

    obs_A = results['obs_strat'][strat_A]['UMSS']
    obs_B = results['obs_strat'][strat_B]['UMSS']
    obs_diff = obs_A - obs_B

    boot_A = results['boot_strat'][strat_A]['UMSS']
    boot_B = results['boot_strat'][strat_B]['UMSS']

    # Only use iterations where both are valid
    valid = ~np.isnan(boot_A) & ~np.isnan(boot_B)
    if np.sum(valid) == 0:
        return None

    diffs = boot_A[valid] - boot_B[valid]
    ci_low = np.percentile(diffs, 2.5)
    ci_high = np.percentile(diffs, 97.5)
    p_val = 2 * min(np.mean(diffs <= 0), np.mean(diffs >= 0))
    p_val = min(p_val, 1.0)

    return {
        'UMSS_A': obs_A, 'UMSS_B': obs_B, 'diff': obs_diff,
        'CI_low': ci_low, 'CI_high': ci_high, 'p_value': p_val,
        'n_valid_boots': int(np.sum(valid)),
    }

thesis_pairs_umss = [
    ("Few-Shot-CoT", "Thought-Experiment"),
    ("Few-Shot-CoT", "Self-Correct"),
    ("Few-Shot", "Thought-Experiment"),
    ("Few-Shot", "Self-Correct"),
    ("Plan-And-Solve", "Thought-Experiment"),
    ("Plan-And-Solve", "Self-Correct"),
    ("Role-Prompting-Proper", "Zero-Shot"),
    ("Few-Shot-CoT", "Few-Shot"),
    ("Few-Shot-CoT", "Role-Prompting-Proper"),
    ("Few-Shot", "Role-Prompting-Proper"),
]

print("\n" + "=" * 100)
print("UMSS PAIRWISE STRATEGY SIGNIFICANCE (Bootstrap)")
print("=" * 100)

strat_sig_records = []
strat_p_values = []

for strat_A, strat_B in thesis_pairs_umss:
    result = bootstrap_pairwise_strategy(results, strat_A, strat_B)
    if result is None:
        print(f"\n{strat_A} vs {strat_B}: insufficient data.")
        continue

    winner = strat_A if result['diff'] > 0 else strat_B
    print(f"\n--- {strat_A} vs {strat_B} ({result['n_valid_boots']} boots) ---")
    print(f"  UMSS({strat_A[:22]}): {result['UMSS_A']:.4f}")
    print(f"  UMSS({strat_B[:22]}): {result['UMSS_B']:.4f}")
    print(f"  Diff: {result['diff']:+.4f}  95% CI: [{result['CI_low']:+.4f}, {result['CI_high']:+.4f}]")
    print(f"  p = {result['p_value']:.5f}  (Winner: {winner})")

    strat_sig_records.append({
        'Strategy_A': strat_A, 'Strategy_B': strat_B,
        'UMSS_A': result['UMSS_A'], 'UMSS_B': result['UMSS_B'],
        'Diff': result['diff'],
        'CI_95': f"[{result['CI_low']:+.4f}, {result['CI_high']:+.4f}]",
        'p_uncorrected': result['p_value'],
        'Winner': winner
    })
    strat_p_values.append(result['p_value'])

if strat_p_values:
    corrected_p = holm_bonferroni_correction(strat_p_values)
    print("\n" + "-" * 100)
    print("HOLM-BONFERRONI CORRECTED (Strategy)")
    print("-" * 100)
    for i, rec in enumerate(strat_sig_records):
        rec['p_corrected'] = corrected_p[i]
        rec['Significant_corrected'] = corrected_p[i] < 0.05
        rec['Significant_uncorrected'] = rec['p_uncorrected'] < 0.05
        sig = "***" if rec['Significant_corrected'] else "n.s."
        print(f"  {rec['Strategy_A']:25s} vs {rec['Strategy_B']:25s} | "
              f"p={rec['p_uncorrected']:.5f} -> p_corr={corrected_p[i]:.5f} {sig}")

pd.DataFrame(strat_sig_records).to_csv("UMSS_Strategy_Pairwise_Significance.csv", index=False)


# ============================================================
# PAIRWISE MODEL COMPARISONS (via model-level bootstrap)
# ============================================================

def bootstrap_pairwise_model(results, model_A, model_B):
    """Compare two models using the model-level bootstrap distributions."""
    if model_A not in results['obs_model'] or model_B not in results['obs_model']:
        return None

    obs_A = results['obs_model'][model_A]['UMSS']
    obs_B = results['obs_model'][model_B]['UMSS']
    obs_diff = obs_A - obs_B

    boot_A = results['boot_model'][model_A]['UMSS']
    boot_B = results['boot_model'][model_B]['UMSS']

    valid = ~np.isnan(boot_A) & ~np.isnan(boot_B)
    if np.sum(valid) == 0:
        return None

    diffs = boot_A[valid] - boot_B[valid]
    ci_low = np.percentile(diffs, 2.5)
    ci_high = np.percentile(diffs, 97.5)
    p_val = 2 * min(np.mean(diffs <= 0), np.mean(diffs >= 0))
    p_val = min(p_val, 1.0)

    return {
        'UMSS_A': obs_A, 'UMSS_B': obs_B, 'diff': obs_diff,
        'CI_low': ci_low, 'CI_high': ci_high, 'p_value': p_val,
        'n_valid_boots': int(np.sum(valid)),
    }

model_pairs_list = list(combinations(UMSS_MODELS, 2))

print("\n" + "=" * 100)
print("UMSS PAIRWISE MODEL SIGNIFICANCE (Bootstrap)")
print("=" * 100)

model_sig_records = []
model_p_values = []

for model_A, model_B in model_pairs_list:
    result = bootstrap_pairwise_model(results, model_A, model_B)
    if result is None:
        continue

    winner = model_A if result['diff'] > 0 else model_B
    print(f"\n--- {model_A} vs {model_B} ({result['n_valid_boots']} boots) ---")
    print(f"  UMSS({model_A}): {result['UMSS_A']:.4f}")
    print(f"  UMSS({model_B}): {result['UMSS_B']:.4f}")
    print(f"  Diff: {result['diff']:+.4f}  95% CI: [{result['CI_low']:+.4f}, {result['CI_high']:+.4f}]")
    print(f"  p = {result['p_value']:.5f}  (Winner: {winner})")

    model_sig_records.append({
        'Model_A': model_A, 'Model_B': model_B,
        'UMSS_A': result['UMSS_A'], 'UMSS_B': result['UMSS_B'],
        'Diff': result['diff'],
        'CI_95': f"[{result['CI_low']:+.4f}, {result['CI_high']:+.4f}]",
        'p_uncorrected': result['p_value'],
        'Winner': winner
    })
    model_p_values.append(result['p_value'])

if model_p_values:
    corrected_p_m = holm_bonferroni_correction(model_p_values)
    print("\n" + "-" * 100)
    print("HOLM-BONFERRONI CORRECTED (Model)")
    print("-" * 100)
    for i, rec in enumerate(model_sig_records):
        rec['p_corrected'] = corrected_p_m[i]
        rec['Significant_corrected'] = corrected_p_m[i] < 0.05
        rec['Significant_uncorrected'] = rec['p_uncorrected'] < 0.05
        sig = "***" if rec['Significant_corrected'] else "n.s."
        print(f"  {rec['Model_A']:12s} vs {rec['Model_B']:12s} | "
              f"p={rec['p_uncorrected']:.5f} -> p_corr={corrected_p_m[i]:.5f} {sig}")

pd.DataFrame(model_sig_records).to_csv("UMSS_Model_Pairwise_Significance.csv", index=False)


# ============================================================
# FULL (MODEL, STRATEGY)-LEVEL TABLE
# ============================================================
records_full = []
for m in UMSS_MODELS:
    for s in UMSS_STRATEGIES:
        p = (m, s)
        if p in results['obs_pair'] and p in results['boot_pair']:
            obs_mcs = results['obs_pair'][p]['MCS']
            obs_srs = results['obs_pair'][p]['SRS']
            if np.isnan(obs_mcs) or np.isnan(obs_srs) or (BETA**2 * obs_mcs + obs_srs) == 0:
                obs_umss = np.nan
            else:
                obs_umss = (1 + BETA**2) * (obs_mcs * obs_srs) / (BETA**2 * obs_mcs + obs_srs)
            bp = results['boot_pair'][p]
            records_full.append({
                'Model': m, 'Strategy': s,
                'MCS': obs_mcs, 'SRS': obs_srs, 'UMSS': obs_umss,
                'UMSS_CI_Low': np.nanpercentile(bp['UMSS'], 2.5),
                'UMSS_CI_High': np.nanpercentile(bp['UMSS'], 97.5),
            })

df_full = pd.DataFrame(records_full).sort_values(['Model', 'UMSS'], ascending=[True, False])
df_full.to_csv("UMSS_Full_ModelStrategy_CIs.csv", index=False)

print("\n" + "=" * 100)
print("FULL (Model, Strategy)-Level UMSS (top 3 per model)")
print("=" * 100)
for m in UMSS_MODELS:
    sub = df_full[df_full['Model'] == m].head(3)
    print(f"\n--- {m} ---")
    for _, row in sub.iterrows():
        print(f"  {row['Strategy']:25s}  UMSS={row['UMSS']:.3f} [{row['UMSS_CI_Low']:.3f}, {row['UMSS_CI_High']:.3f}]")


# ============================================================
# SUMMARY
# ============================================================
print("\n" + "=" * 80)
print("SUMMARY")
print("=" * 80)
if strat_sig_records:
    n_unc = sum(1 for r in strat_sig_records if r.get('Significant_uncorrected', False))
    n_cor = sum(1 for r in strat_sig_records if r.get('Significant_corrected', False))
    print(f"Strategy: {n_unc}/{len(strat_sig_records)} sig uncorrected, {n_cor}/{len(strat_sig_records)} after Holm-Bonferroni")
if model_sig_records:
    n_unc_m = sum(1 for r in model_sig_records if r.get('Significant_uncorrected', False))
    n_cor_m = sum(1 for r in model_sig_records if r.get('Significant_corrected', False))
    print(f"Model:    {n_unc_m}/{len(model_sig_records)} sig uncorrected, {n_cor_m}/{len(model_sig_records)} after Holm-Bonferroni")
```

```python
########################################################################
# UMSS STATISTICAL SIGNIFICANCE - SINGLE SELF-CONTAINED CELL
# Paste this after all existing cells and run.
# Requires: `data` dict from Cell 3/4 to already exist.
########################################################################

import numpy as np
import pandas as pd
from itertools import combinations

# ==================================================================
# STEP 0: Build ETHICS_Contrast_FlipAccuracy from raw contrast data
# ==================================================================
print("STEP 0: Building ETHICS_Contrast_FlipAccuracy...")
print(f"  Current data keys: {list(data.keys())}")

data["ETHICS_Contrast_FlipAccuracy"] = {}

# The contrast data lives under "ETHICS - Contrast" with keys like
# "Few-Shot - Originals" and "Few-Shot - Contrasts"
src_key = "ETHICS - Contrast"
if src_key in data:
    for model in data[src_key]:
        data["ETHICS_Contrast_FlipAccuracy"][model] = {}
        for raw_strat_name, arr in data[src_key][model].items():
            if raw_strat_name.endswith(" - Contrasts"):
                base = raw_strat_name.replace(" - Contrasts", "")
                data["ETHICS_Contrast_FlipAccuracy"][model][base] = arr
    # Print what we got
    for m in data["ETHICS_Contrast_FlipAccuracy"]:
        strats = sorted(data["ETHICS_Contrast_FlipAccuracy"][m].keys())
        counts = {s: int(np.sum(~np.isnan(data["ETHICS_Contrast_FlipAccuracy"][m][s]))) for s in strats}
        print(f"  {m}: {counts}")
else:
    print(f"  *** ERROR: '{src_key}' not found in data dict! ***")
    print(f"  Available: {list(data.keys())}")

print(f"  Updated data keys: {list(data.keys())}")
print()

# ==================================================================
# STEP 1: UMSS Engine (matching paper methodology exactly)
# ==================================================================
MCS_DATASETS = ["ETHICS", "Scruples", "ETHICS_Contrast_FlipAccuracy"]
SRS_DATASETS = ["WildJailbreak_Harmful", "WildJailbreak_Benign"]
ALL_DS = MCS_DATASETS + SRS_DATASETS

UMSS_MODELS = ["GPT-4.1", "Sonnet-4", "Gemini 2.5", "Deepseek"]
UMSS_STRATEGIES = [
    "Few-Shot", "Few-Shot-CoT", "First-Principles", "Plan-And-Solve",
    "Role-Prompting-Proper", "Self-Correct", "Thought-Experiment",
    "Value-Grounded", "Zero-Shot", "Zero-Shot_Cot"
]

# Verify all 5 datasets exist
for ds in ALL_DS:
    if ds in data:
        n = sum(len(data[ds][m]) for m in data[ds])
        print(f"  {ds}: {n} (model,strategy) pairs")
    else:
        print(f"  *** {ds}: MISSING ***")
print()


def _minmax(values):
    mn, mx = min(values), max(values)
    if mx == mn:
        return [0.5] * len(values)
    return [(v - mn) / (mx - mn) for v in values]


def compute_raw_means(data_dict, resample=False, rng=None):
    """Get raw accuracy/rate for every (model, strategy, dataset)."""
    raw = {}
    for m in UMSS_MODELS:
        for s in UMSS_STRATEGIES:
            raw[(m, s)] = {}
            for ds in ALL_DS:
                if ds in data_dict and m in data_dict[ds] and s in data_dict[ds][m]:
                    arr = data_dict[ds][m][s]
                    valid = arr[~np.isnan(arr)]
                    if len(valid) == 0:
                        continue
                    if resample and rng is not None:
                        valid = rng.choice(valid, size=len(valid), replace=True)
                    raw[(m, s)][ds] = float(np.mean(valid))
    return raw


def compute_strategy_level_umss(raw_means, beta=1.0):
    """
    Table 6 methodology:
    1) Average raw scores across 4 models for each strategy
    2) Min-max normalize across the 10 strategy averages
    3) MCS = mean of 3 normalized MCS metrics
    4) SRS = mean of 2 normalized SRS metrics
    5) UMSS = harmonic mean
    """
    # Average across models
    strat_raw = {}
    for s in UMSS_STRATEGIES:
        strat_raw[s] = {}
        for ds in ALL_DS:
            vals = [raw_means[(m, s)].get(ds) for m in UMSS_MODELS]
            vals = [v for v in vals if v is not None]
            if vals:
                strat_raw[s][ds] = np.mean(vals)

    # Normalize across 10 strategies
    strat_norm = {s: {} for s in UMSS_STRATEGIES}
    for ds in ALL_DS:
        ds_vals = [(s, strat_raw[s].get(ds)) for s in UMSS_STRATEGIES]
        valid = [(s, v) for s, v in ds_vals if v is not None]
        if len(valid) < 2:
            for s in UMSS_STRATEGIES:
                strat_norm[s][ds] = 0.5
            continue
        normed = _minmax([v for _, v in valid])
        for (s, _), n in zip(valid, normed):
            strat_norm[s][ds] = n

    # MCS, SRS, UMSS
    out = {}
    for s in UMSS_STRATEGIES:
        mcs_v = [strat_norm[s].get(ds) for ds in MCS_DATASETS]
        mcs_v = [v for v in mcs_v if v is not None]
        mcs = np.mean(mcs_v) if mcs_v else np.nan

        srs_v = [strat_norm[s].get(ds) for ds in SRS_DATASETS]
        srs_v = [v for v in srs_v if v is not None]
        srs = np.mean(srs_v) if srs_v else np.nan

        if np.isnan(mcs) or np.isnan(srs) or (beta**2 * mcs + srs) == 0:
            umss = np.nan
        else:
            umss = (1 + beta**2) * (mcs * srs) / (beta**2 * mcs + srs)
        out[s] = {"MCS": mcs, "SRS": srs, "UMSS": umss}
    return out


def compute_model_level_umss(raw_means, beta=1.0):
    """
    Table 5 methodology:
    1) Min-max normalize each metric across all 40 (model, strategy) pairs
    2) Compute MCS/SRS per pair
    3) Average MCS/SRS by model
    4) UMSS = harmonic mean of model-level MCS and SRS
    """
    pairs = [(m, s) for m in UMSS_MODELS for s in UMSS_STRATEGIES]

    pair_norm = {p: {} for p in pairs}
    for ds in ALL_DS:
        ds_vals = [(p, raw_means[p].get(ds)) for p in pairs]
        valid = [(p, v) for p, v in ds_vals if v is not None]
        if len(valid) < 2:
            for p in pairs:
                pair_norm[p][ds] = 0.5
            continue
        normed = _minmax([v for _, v in valid])
        for (p, _), n in zip(valid, normed):
            pair_norm[p][ds] = n

    pair_scores = {}
    for p in pairs:
        mcs_v = [pair_norm[p].get(ds) for ds in MCS_DATASETS]
        mcs_v = [v for v in mcs_v if v is not None]
        mcs = np.mean(mcs_v) if mcs_v else np.nan
        srs_v = [pair_norm[p].get(ds) for ds in SRS_DATASETS]
        srs_v = [v for v in srs_v if v is not None]
        srs = np.mean(srs_v) if srs_v else np.nan
        pair_scores[p] = {"MCS": mcs, "SRS": srs}

    out = {}
    for m in UMSS_MODELS:
        mcs_all = [pair_scores[(m, s)]["MCS"] for s in UMSS_STRATEGIES]
        srs_all = [pair_scores[(m, s)]["SRS"] for s in UMSS_STRATEGIES]
        avg_mcs = np.nanmean(mcs_all)
        avg_srs = np.nanmean(srs_all)
        if np.isnan(avg_mcs) or np.isnan(avg_srs) or (beta**2 * avg_mcs + avg_srs) == 0:
            umss = np.nan
        else:
            umss = (1 + beta**2) * (avg_mcs * avg_srs) / (beta**2 * avg_mcs + avg_srs)
        out[m] = {"MCS": avg_mcs, "SRS": avg_srs, "UMSS": umss}
    return out


# ==================================================================
# STEP 2: Run bootstrap
# ==================================================================
N_BOOTSTRAPS = 10000
BETA = 1.0

rng = np.random.default_rng(42)
raw_obs = compute_raw_means(data, resample=False)
obs_strat = compute_strategy_level_umss(raw_obs, BETA)
obs_model = compute_model_level_umss(raw_obs, BETA)

boot_strat = {s: {"MCS": [], "SRS": [], "UMSS": []} for s in UMSS_STRATEGIES}
boot_model = {m: {"MCS": [], "SRS": [], "UMSS": []} for m in UMSS_MODELS}

print(f"Running {N_BOOTSTRAPS} bootstrap iterations...")
for b in range(N_BOOTSTRAPS):
    raw_b = compute_raw_means(data, resample=True, rng=rng)
    sb = compute_strategy_level_umss(raw_b, BETA)
    mb = compute_model_level_umss(raw_b, BETA)
    for s in UMSS_STRATEGIES:
        for k in ["MCS", "SRS", "UMSS"]:
            boot_strat[s][k].append(sb[s][k])
    for m in UMSS_MODELS:
        for k in ["MCS", "SRS", "UMSS"]:
            boot_model[m][k].append(mb[m][k])

for d in [boot_strat, boot_model]:
    for key in d:
        for metric in ["MCS", "SRS", "UMSS"]:
            d[key][metric] = np.array(d[key][metric])

print("Done.\n")


# ==================================================================
# STEP 3: Print tables
# ==================================================================
print("=" * 90)
print("TABLE 5: Model-Level UMSS with Bootstrap 95% CIs")
print("=" * 90)
rows5 = []
for m in UMSS_MODELS:
    o = obs_model[m]
    b = boot_model[m]
    rows5.append({
        "Model": m, "MCS": o["MCS"],
        "MCS_CI": f'[{np.nanpercentile(b["MCS"], 2.5):.3f}, {np.nanpercentile(b["MCS"], 97.5):.3f}]',
        "SRS": o["SRS"],
        "SRS_CI": f'[{np.nanpercentile(b["SRS"], 2.5):.3f}, {np.nanpercentile(b["SRS"], 97.5):.3f}]',
        "UMSS": o["UMSS"],
        "UMSS_CI": f'[{np.nanpercentile(b["UMSS"], 2.5):.3f}, {np.nanpercentile(b["UMSS"], 97.5):.3f}]',
    })
df5 = pd.DataFrame(rows5).sort_values("UMSS", ascending=False)
print(df5.to_string(index=False))
df5.to_csv("UMSS_Model_Level_CIs.csv", index=False)

print("\n" + "=" * 100)
print("TABLE 6: Strategy-Level UMSS with Bootstrap 95% CIs")
print("=" * 100)
rows6 = []
for s in UMSS_STRATEGIES:
    o = obs_strat[s]
    b = boot_strat[s]
    rows6.append({
        "Strategy": s, "MCS": o["MCS"],
        "MCS_CI": f'[{np.nanpercentile(b["MCS"], 2.5):.3f}, {np.nanpercentile(b["MCS"], 97.5):.3f}]',
        "SRS": o["SRS"],
        "SRS_CI": f'[{np.nanpercentile(b["SRS"], 2.5):.3f}, {np.nanpercentile(b["SRS"], 97.5):.3f}]',
        "UMSS": o["UMSS"],
        "UMSS_CI": f'[{np.nanpercentile(b["UMSS"], 2.5):.3f}, {np.nanpercentile(b["UMSS"], 97.5):.3f}]',
    })
df6 = pd.DataFrame(rows6).sort_values("UMSS", ascending=False)
print(df6.to_string(index=False))
df6.to_csv("UMSS_Strategy_Level_CIs.csv", index=False)


# ==================================================================
# STEP 4: Pairwise strategy comparisons + Holm-Bonferroni
# ==================================================================
def _pairwise_boot(bA, bB):
    valid = ~np.isnan(bA) & ~np.isnan(bB)
    if np.sum(valid) == 0:
        return None
    diffs = bA[valid] - bB[valid]
    p = 2 * min(np.mean(diffs <= 0), np.mean(diffs >= 0))
    return {"diffs": diffs, "p": min(p, 1.0),
            "ci_lo": np.percentile(diffs, 2.5), "ci_hi": np.percentile(diffs, 97.5)}

thesis_pairs = [
    ("Few-Shot-CoT", "Thought-Experiment"), ("Few-Shot-CoT", "Self-Correct"),
    ("Few-Shot", "Thought-Experiment"), ("Few-Shot", "Self-Correct"),
    ("Plan-And-Solve", "Thought-Experiment"), ("Plan-And-Solve", "Self-Correct"),
    ("Role-Prompting-Proper", "Zero-Shot"),
    ("Few-Shot-CoT", "Few-Shot"), ("Few-Shot-CoT", "Role-Prompting-Proper"),
    ("Few-Shot", "Role-Prompting-Proper"),
]

print("\n" + "=" * 100)
print("PAIRWISE STRATEGY SIGNIFICANCE")
print("=" * 100)
srecs, sp = [], []
for sA, sB in thesis_pairs:
    r = _pairwise_boot(boot_strat[sA]["UMSS"], boot_strat[sB]["UMSS"])
    if r is None:
        continue
    oA, oB = obs_strat[sA]["UMSS"], obs_strat[sB]["UMSS"]
    diff = oA - oB
    w = sA if diff > 0 else sB
    print(f"\n  {sA} vs {sB}")
    print(f"    UMSS: {oA:.4f} vs {oB:.4f}, diff={diff:+.4f}, 95%CI=[{r['ci_lo']:+.4f},{r['ci_hi']:+.4f}], p={r['p']:.5f}")
    srecs.append({"A": sA, "B": sB, "diff": diff, "p": r["p"], "W": w})
    sp.append(r["p"])

# Holm-Bonferroni
if sp:
    n = len(sp)
    indexed = sorted(range(n), key=lambda i: sp[i])
    cp = [0.0] * n
    mx = 0.0
    for rank, i in enumerate(indexed):
        adj = min(sp[i] * (n - rank), 1.0)
        mx = max(mx, adj)
        cp[i] = mx
    print("\n  Holm-Bonferroni corrected:")
    for i, rec in enumerate(srecs):
        sig = "***" if cp[i] < 0.05 else "n.s."
        rec["p_corr"] = cp[i]
        print(f"    {rec['A']:25s} vs {rec['B']:25s} | p={rec['p']:.5f} -> p_corr={cp[i]:.5f} {sig}")
    pd.DataFrame(srecs).to_csv("UMSS_Strategy_Pairwise_Significance.csv", index=False)


# ==================================================================
# STEP 5: Pairwise model comparisons + Holm-Bonferroni
# ==================================================================
print("\n" + "=" * 100)
print("PAIRWISE MODEL SIGNIFICANCE")
print("=" * 100)
mrecs, mp = [], []
for mA, mB in combinations(UMSS_MODELS, 2):
    r = _pairwise_boot(boot_model[mA]["UMSS"], boot_model[mB]["UMSS"])
    if r is None:
        continue
    oA, oB = obs_model[mA]["UMSS"], obs_model[mB]["UMSS"]
    diff = oA - oB
    w = mA if diff > 0 else mB
    print(f"\n  {mA} vs {mB}")
    print(f"    UMSS: {oA:.4f} vs {oB:.4f}, diff={diff:+.4f}, 95%CI=[{r['ci_lo']:+.4f},{r['ci_hi']:+.4f}], p={r['p']:.5f}")
    mrecs.append({"A": mA, "B": mB, "diff": diff, "p": r["p"], "W": w})
    mp.append(r["p"])

if mp:
    n = len(mp)
    indexed = sorted(range(n), key=lambda i: mp[i])
    cp = [0.0] * n
    mx = 0.0
    for rank, i in enumerate(indexed):
        adj = min(mp[i] * (n - rank), 1.0)
        mx = max(mx, adj)
        cp[i] = mx
    print("\n  Holm-Bonferroni corrected:")
    for i, rec in enumerate(mrecs):
        sig = "***" if cp[i] < 0.05 else "n.s."
        rec["p_corr"] = cp[i]
        print(f"    {rec['A']:12s} vs {rec['B']:12s} | p={rec['p']:.5f} -> p_corr={cp[i]:.5f} {sig}")
    pd.DataFrame(mrecs).to_csv("UMSS_Model_Pairwise_Significance.csv", index=False)

print("\n" + "=" * 80)
print("DONE. CSV files saved.")
print("=" * 80)
```

```python
########################################################################
# UMSS ABLATION ANALYSES
# Run AFTER the main UMSS cell (requires `data` dict and helper funcs)
# Produces:
#   1) Outlier removal robustness check (Appendix table)
#   2) Aggregation ablation table (arithmetic, geometric, harmonic, weighted-HM)
########################################################################

import numpy as np
import pandas as pd

# ---- Uses same constants from main cell ----
# UMSS_MODELS, UMSS_STRATEGIES, MCS_DATASETS, SRS_DATASETS, ALL_DS,
# compute_raw_means, _minmax  — all assumed to exist already.

# ==================================================================
# ABLATION 1: Outlier Removal Robustness
# Drop the single best and worst (model, strategy) pair from the
# normalization pool and check whether strategy rankings change.
# ==================================================================

def compute_strategy_umss_excluding(raw_means, exclude_pairs, beta=1.0):
    """
    Same as compute_strategy_level_umss but excludes specified
    (model, strategy) pairs from min-max normalization bounds.
    """
    # Average across models (only non-excluded)
    strat_raw = {}
    for s in UMSS_STRATEGIES:
        strat_raw[s] = {}
        for ds in ALL_DS:
            vals = []
            for m in UMSS_MODELS:
                if (m, s) in exclude_pairs:
                    continue
                v = raw_means[(m, s)].get(ds)
                if v is not None:
                    vals.append(v)
            if vals:
                strat_raw[s][ds] = np.mean(vals)

    # Min-max normalize across strategies
    strat_norm = {s: {} for s in UMSS_STRATEGIES}
    for ds in ALL_DS:
        ds_vals = [(s, strat_raw[s].get(ds)) for s in UMSS_STRATEGIES]
        valid = [(s, v) for s, v in ds_vals if v is not None]
        if len(valid) < 2:
            for s in UMSS_STRATEGIES:
                strat_norm[s][ds] = 0.5
            continue
        normed = _minmax([v for _, v in valid])
        for (s, _), n in zip(valid, normed):
            strat_norm[s][ds] = n

    out = {}
    for s in UMSS_STRATEGIES:
        mcs_v = [strat_norm[s].get(ds) for ds in MCS_DATASETS if strat_norm[s].get(ds) is not None]
        srs_v = [strat_norm[s].get(ds) for ds in SRS_DATASETS if strat_norm[s].get(ds) is not None]
        mcs = np.mean(mcs_v) if mcs_v else np.nan
        srs = np.mean(srs_v) if srs_v else np.nan
        if np.isnan(mcs) or np.isnan(srs) or (beta**2 * mcs + srs) == 0:
            umss = np.nan
        else:
            umss = (1 + beta**2) * (mcs * srs) / (beta**2 * mcs + srs)
        out[s] = umss
    return out


print("=" * 90)
print("ABLATION 1: Outlier Removal Robustness Check")
print("=" * 90)

raw_obs = compute_raw_means(data, resample=False)

# Baseline ranking
baseline_umss = compute_strategy_umss_excluding(raw_obs, set(), beta=1.0)
baseline_ranking = sorted(UMSS_STRATEGIES, key=lambda s: -baseline_umss[s])

# Find the best and worst (model, strategy) pairs by overall raw accuracy
# across all datasets (to identify the outlier pairs)
pair_overall = {}
for m in UMSS_MODELS:
    for s in UMSS_STRATEGIES:
        vals = [raw_obs[(m, s)].get(ds) for ds in ALL_DS]
        vals = [v for v in vals if v is not None]
        if vals:
            pair_overall[(m, s)] = np.mean(vals)

sorted_pairs = sorted(pair_overall.items(), key=lambda x: x[1])
worst_pair = sorted_pairs[0][0]
best_pair = sorted_pairs[-1][0]

print(f"\nBaseline strategy ranking (by UMSS):")
for i, s in enumerate(baseline_ranking):
    print(f"  {i+1}. {s:30s} UMSS={baseline_umss[s]:.4f}")

# Remove best
umss_no_best = compute_strategy_umss_excluding(raw_obs, {best_pair}, beta=1.0)
rank_no_best = sorted(UMSS_STRATEGIES, key=lambda s: -umss_no_best[s])

# Remove worst
umss_no_worst = compute_strategy_umss_excluding(raw_obs, {worst_pair}, beta=1.0)
rank_no_worst = sorted(UMSS_STRATEGIES, key=lambda s: -umss_no_worst[s])

# Remove both
umss_no_both = compute_strategy_umss_excluding(raw_obs, {best_pair, worst_pair}, beta=1.0)
rank_no_both = sorted(UMSS_STRATEGIES, key=lambda s: -umss_no_both[s])

print(f"\nExcluded best pair:  {best_pair}  (mean raw = {pair_overall[best_pair]:.4f})")
print(f"Excluded worst pair: {worst_pair} (mean raw = {pair_overall[worst_pair]:.4f})")

# Compute max rank displacement
def max_displacement(base_rank, test_rank):
    base_pos = {s: i for i, s in enumerate(base_rank)}
    return max(abs(base_pos[s] - i) for i, s in enumerate(test_rank))

d_best = max_displacement(baseline_ranking, rank_no_best)
d_worst = max_displacement(baseline_ranking, rank_no_worst)
d_both = max_displacement(baseline_ranking, rank_no_both)

print(f"\nMax rank displacement:")
print(f"  Remove best:       {d_best} position(s)")
print(f"  Remove worst:      {d_worst} position(s)")
print(f"  Remove both:       {d_both} position(s)")

# Build table
rows_outlier = []
for i, s in enumerate(baseline_ranking):
    pos_nb = rank_no_best.index(s) + 1
    pos_nw = rank_no_worst.index(s) + 1
    pos_nboth = rank_no_both.index(s) + 1
    rows_outlier.append({
        "Strategy": s,
        "Baseline_Rank": i + 1,
        "Baseline_UMSS": round(baseline_umss[s], 4),
        "No_Best_Rank": pos_nb,
        "No_Best_UMSS": round(umss_no_best[s], 4),
        "No_Worst_Rank": pos_nw,
        "No_Worst_UMSS": round(umss_no_worst[s], 4),
        "No_Both_Rank": pos_nboth,
        "No_Both_UMSS": round(umss_no_both[s], 4),
    })

df_outlier = pd.DataFrame(rows_outlier)
print("\n" + df_outlier.to_string(index=False))
df_outlier.to_csv("UMSS_Outlier_Robustness.csv", index=False)


# ==================================================================
# ABLATION 2: Aggregation Method Comparison
# Compare UMSS under: Harmonic, Arithmetic, Geometric, Weighted-HM
# ==================================================================

print("\n\n" + "=" * 90)
print("ABLATION 2: Aggregation Method Comparison")
print("=" * 90)


def aggregate_mcs_srs(mcs, srs, method="harmonic", beta=1.0):
    if np.isnan(mcs) or np.isnan(srs):
        return np.nan
    if method == "harmonic":
        denom = beta**2 * mcs + srs
        if denom == 0:
            return 0.0
        return (1 + beta**2) * (mcs * srs) / denom
    elif method == "arithmetic":
        return (mcs + srs) / 2.0
    elif method == "geometric":
        if mcs <= 0 or srs <= 0:
            return 0.0
        return np.sqrt(mcs * srs)
    elif method == "weighted_harmonic":
        # Downweight near-saturated ETHICS in MCS computation
        # We handle this separately below
        raise ValueError("weighted_harmonic handled externally")
    else:
        raise ValueError(f"Unknown method: {method}")


def compute_strategy_umss_with_method(raw_means, method="harmonic", beta=1.0, mcs_weights=None):
    """
    Like compute_strategy_level_umss but with configurable aggregation.
    mcs_weights: dict mapping dataset -> weight for MCS averaging.
    """
    # Average across models
    strat_raw = {}
    for s in UMSS_STRATEGIES:
        strat_raw[s] = {}
        for ds in ALL_DS:
            vals = [raw_means[(m, s)].get(ds) for m in UMSS_MODELS]
            vals = [v for v in vals if v is not None]
            if vals:
                strat_raw[s][ds] = np.mean(vals)

    # Min-max normalize
    strat_norm = {s: {} for s in UMSS_STRATEGIES}
    for ds in ALL_DS:
        ds_vals = [(s, strat_raw[s].get(ds)) for s in UMSS_STRATEGIES]
        valid = [(s, v) for s, v in ds_vals if v is not None]
        if len(valid) < 2:
            for s in UMSS_STRATEGIES:
                strat_norm[s][ds] = 0.5
            continue
        normed = _minmax([v for _, v in valid])
        for (s, _), n in zip(valid, normed):
            strat_norm[s][ds] = n

    # MCS with optional weighting
    if mcs_weights is None:
        mcs_weights = {ds: 1.0 for ds in MCS_DATASETS}

    out = {}
    for s in UMSS_STRATEGIES:
        # Weighted MCS
        mcs_vals = []
        mcs_ws = []
        for ds in MCS_DATASETS:
            v = strat_norm[s].get(ds)
            if v is not None:
                mcs_vals.append(v)
                mcs_ws.append(mcs_weights.get(ds, 1.0))
        if mcs_vals:
            mcs = np.average(mcs_vals, weights=mcs_ws)
        else:
            mcs = np.nan

        # SRS (always equal weight)
        srs_v = [strat_norm[s].get(ds) for ds in SRS_DATASETS if strat_norm[s].get(ds) is not None]
        srs = np.mean(srs_v) if srs_v else np.nan

        umss = aggregate_mcs_srs(mcs, srs, method=method, beta=beta)
        out[s] = {"MCS": mcs, "SRS": srs, "UMSS": umss}
    return out


methods = {
    "Harmonic": {"method": "harmonic", "weights": None},
    "Arithmetic": {"method": "arithmetic", "weights": None},
    "Geometric": {"method": "geometric", "weights": None},
    "Weighted-HM": {
        "method": "harmonic",
        "weights": {
            "ETHICS": 0.5,                          # downweight saturated task
            "ETHICS_Contrast_FlipAccuracy": 1.0,
            "Scruples": 1.0,
        }
    },
}

agg_rows = []
agg_rankings = {}
for label, cfg in methods.items():
    result = compute_strategy_umss_with_method(
        raw_obs, method=cfg["method"], beta=1.0, mcs_weights=cfg["weights"]
    )
    ranking = sorted(UMSS_STRATEGIES, key=lambda s: -result[s]["UMSS"])
    agg_rankings[label] = ranking
    for s in UMSS_STRATEGIES:
        agg_rows.append({
            "Strategy": s,
            "Method": label,
            "MCS": round(result[s]["MCS"], 3),
            "SRS": round(result[s]["SRS"], 3),
            "UMSS": round(result[s]["UMSS"], 3),
            "Rank": ranking.index(s) + 1,
        })

df_agg = pd.DataFrame(agg_rows)

# Pivot for display
pivot = df_agg.pivot(index="Strategy", columns="Method", values="UMSS")
pivot = pivot[["Harmonic", "Arithmetic", "Geometric", "Weighted-HM"]]
# Add rank columns
for label in ["Harmonic", "Arithmetic", "Geometric", "Weighted-HM"]:
    rank_col = df_agg[df_agg["Method"] == label].set_index("Strategy")["Rank"]
    pivot[f"{label}_Rank"] = rank_col

pivot = pivot.sort_values("Harmonic", ascending=False)
print("\nStrategy UMSS under different aggregation methods:")
print(pivot.to_string())
pivot.to_csv("UMSS_Aggregation_Ablation.csv")

# Check top-3 stability
print("\n\nTop-3 rankings by method:")
for label, ranking in agg_rankings.items():
    top3 = ranking[:3]
    print(f"  {label:15s}: {', '.join(top3)}")

# Max rank displacement across methods vs harmonic baseline
harm_ranking = agg_rankings["Harmonic"]
harm_pos = {s: i for i, s in enumerate(harm_ranking)}
for label in ["Arithmetic", "Geometric", "Weighted-HM"]:
    other = agg_rankings[label]
    d = max(abs(harm_pos[s] - i) for i, s in enumerate(other))
    print(f"  Max displacement {label} vs Harmonic: {d} position(s)")

print("\n" + "=" * 80)
print("DONE. CSVs saved: UMSS_Outlier_Robustness.csv, UMSS_Aggregation_Ablation.csv")
print("=" * 80)
```
